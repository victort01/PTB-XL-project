# 02 - Analise exploratoria

**Objetivo:** gerar figuras e tabelas iniciais para o TCC.

As figuras sao salvas em `reports/figures` em PNG e PDF quando aplicavel.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().resolve()
if str(PROJECT_ROOT / 'src') not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT / 'src'))

from tcc_ecg.config import load_config
from tcc_ecg.data import prepare_metadata, final_labeled_records
from tcc_ecg.paths import resolve_project_path
from tcc_ecg.plots import (
    plot_age_distribution,
    plot_class_distribution,
    plot_example_signals_by_class,
    plot_records_by_fold,
)

config = load_config()
metadata = prepare_metadata(config, save_summary=True)
final_metadata = final_labeled_records(metadata)
figures_dir = resolve_project_path(config['outputs']['figures_dir'], config['project_root'])
```

```python
raw_counts = metadata.explode('diagnostic_superclasses')['diagnostic_superclasses'].value_counts()
plot_class_distribution(raw_counts, 'Distribuicao das superclasses antes da filtragem multiclasse', figures_dir / 'fig_class_distribution_raw.png')

final_counts = final_metadata['target'].value_counts()
plot_class_distribution(final_counts, 'Distribuicao das classes apos estrategia multiclasse', figures_dir / 'fig_class_distribution_final.png')

sex_counts = final_metadata['sex'].value_counts(dropna=False)
plot_class_distribution(sex_counts, 'Distribuicao por sexo', figures_dir / 'fig_sex_distribution.png')

plot_age_distribution(final_metadata, figures_dir / 'fig_age_distribution_clean.png')
plot_records_by_fold(final_metadata, figures_dir / 'fig_records_by_fold.png')
plot_example_signals_by_class(final_metadata, config, figures_dir / 'fig_example_signals_by_class.png')
print('Figuras de EDA geradas em:', figures_dir)
```
