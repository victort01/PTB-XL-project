# 00 - Verificacao do ambiente

**Objetivo:** confirmar imports, configuracao, estrutura de pastas e presenca dos arquivos PTB-XL.

**Entrada:** `configs/config.yaml` e pasta `data/raw/ptb-xl`.

**Saida:** diagnostico inicial do ambiente, sem treinar modelos.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().resolve()
if str(PROJECT_ROOT / 'src') not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT / 'src'))

from tcc_ecg.config import load_config
from tcc_ecg.data import check_ptbxl_files
from tcc_ecg.paths import ensure_dir, resolve_project_path
from tcc_ecg.utils import setup_logging, set_random_seed

setup_logging()
config = load_config()
set_random_seed(config['project']['seed'])
print('Projeto:', config['project_root'])
print('Seed:', config['project']['seed'])
print('Frequencia:', config['data']['signal_frequency'])
```

```python
for key in ['figures_dir', 'tables_dir', 'models_dir', 'processed_dir']:
    path = ensure_dir(resolve_project_path(config['outputs'][key], config['project_root']))
    print(f'{key}: {path}')

try:
    paths = check_ptbxl_files(config)
    print('Arquivos PTB-XL encontrados:')
    for name, path in paths.items():
        print(f'- {name}: {path}')
except FileNotFoundError as exc:
    print(exc)
```
