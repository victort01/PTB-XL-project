# 04 - Treinamento dos modelos classicos

**Objetivo:** treinar Regressao Logistica, SVM, Random Forest, LightGBM e CatBoost.

O fold 9 e usado para escolha de hiperparametros leves. O fold 10 fica reservado para avaliacao final. SMOTE, quando ativado, e aplicado somente no treino.

```python
from pathlib import Path
import sys
import pandas as pd

PROJECT_ROOT = Path.cwd().resolve()
if str(PROJECT_ROOT / 'src') not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT / 'src'))

from tcc_ecg.config import load_config
from tcc_ecg.data import prepare_metadata
from tcc_ecg.features import build_feature_table
from tcc_ecg.models import train_and_evaluate_models
from tcc_ecg.paths import resolve_project_path

config = load_config()
features_path = resolve_project_path(config['outputs']['processed_dir'], config['project_root']) / 'features.parquet'
if features_path.exists():
    features = pd.read_parquet(features_path)
else:
    metadata = prepare_metadata(config, save_summary=True)
    features, _ = build_feature_table(metadata, config, save=True)

results = train_and_evaluate_models(features, config)
metrics = results['metrics'].sort_values(['split', 'f1_macro'], ascending=[True, False])
display(metrics)
```
